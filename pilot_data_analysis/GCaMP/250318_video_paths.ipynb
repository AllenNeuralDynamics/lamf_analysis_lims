```python
import pandas as pd
import numpy as np
from pathlib import Path
```

```python
plane_save_fn = Path(r'\\allen\programs\mindscope\workgroups\learning\pilots\GCaMP8') / 'gcamp_pilot_plane_info_250226.csv'
plane_info = pd.read_csv(plane_save_fn)
run_df_fn = Path(r'D:\GCaMP characterization\suite2p_format_correct_spks\run_df.csv')
run_df = pd.read_csv(run_df_fn)
```

```python
def get_movie_path(row):
    session_dir = row.session_storage_directory
    opid = row.ophys_experiment_id
    plane_dir = Path(session_dir) / f'ophys_experiment_{opid}'
    movie_path = plane_dir / f'processed/{opid}_suite2p_motion_preview.webm'
    return movie_path
```

```python
plane_info['movie_path'] = plane_info.apply(lambda x: get_movie_path(x), axis=1)
plane_info['targeted_depth'] = (plane_info['depth'] // 105) * 100 + 75
plane_info['targeted_depth'] = plane_info['targeted_depth'].astype(int)
plane_info['date'] = plane_info['date_of_acquisition'].str.split('_').str[0]
```

```python
suite2p_dir_list = list(suite2p_dir.glob('*'))
mouse_ids = [int(x.name.split('_')[-3]) for x in suite2p_dir_list]
acquisition_dates = [x.name.split('_')[-2] for x in suite2p_dir_list]
targeted_depths = [int(x.name.split('_')[-1]) for x in suite2p_dir_list]
gcamps = ['_'.join(x.name.split('_')[:-3]) for x in suite2p_dir_list]
suite2p_df = pd.DataFrame({'gcamp': gcamps, 
                           'mouse_id': mouse_ids,
                           'date': acquisition_dates,
                           'targeted_depth': targeted_depths,
                           })
```

```python
def get_movie_path_from_plane_info(row, plane_info):
    plane = plane_info[(plane_info['date'] == row.date) &
                          (plane_info['targeted_depth'] == row.targeted_depth) &
                          (plane_info['mouse_id'] == row.mouse_id)]
    assert len(plane) == 1
    return plane['movie_path'].values[0]
```

```python
merged_df = suite2p_df.merge(plane_info[['mouse_id', 'date', 'targeted_depth', 'movie_path']],
                            on=['mouse_id', 'date', 'targeted_depth'],
                            how='left')
```

```python
save_fn = Path(r'D:\GCaMP characterization\suite2p_format_correct_spks\movie_paths.csv')
merged_df.to_csv(save_fn, index=False)
```
