```python
# Create a dataframe with mouse id and date of acquisition
def get_mouse_id_from_session_id(osid):
    try:
        mouse_id = from_lims_utilities.get_mouse_ids_from_id('specimen_id', from_lims.get_general_info_for_ophys_session_id(osid)['specimen_id'].values[0]).labtracks_id.values[0]
    except:
        print(f'{osid} error')
    return mouse_id

def get_date_of_acquisition_from_session_id(osid):
    return str(from_lims.get_general_info_for_ophys_session_id(osids[0])['date_of_acquisition'].values[0]).split('T')[0]
```

```python
mouse_ids = [get_mouse_id_from_session_id(osid) for osid in osids]
date_of_acquisitions = [get_date_of_acquisition_from_session_id(osid) for osid in osids]
osid_session_key_df = pd.DataFrame({"session_id": osids,
                                    "mouse_id": mouse_ids,
                                    "date_of_acquisition": date_of_acquisitions})
osid_session_key_df['session_key'] = osid_session_key_df.apply(lambda x: x.mouse_id + '_' + x.date_of_acquisition, axis=1)
```
