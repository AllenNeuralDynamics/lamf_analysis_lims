```python
from pathlib import Path
import pandas as pd
import shutil
```

```python
base_dir = Path(r"\\allen\programs\mindscope\workgroups\learning\sam\stitched_fullfields_gcampchar")
table = pd.read_csv(base_dir / "20251112_gcampcharmice.csv")
mouse_ids = table.external_specimen_name.unique()
for mouse_id in mouse_ids:
    mouse_table = table.query('external_specimen_name == @mouse_id')
    mouse_dir = base_dir / f'{mouse_id}'
    mouse_dir.mkdir(exist_ok=True, parents=True)
    session_ids = mouse_table.session_id.unique()
    for session_id in session_ids:
        try:
            fn = next(base_dir.glob(f'{session_id}_fullfieldROImatch.png'))
            shutil.copy(fn, mouse_dir)
        except StopIteration:
            pass
        
```
